```python
import sympy as sp
t = sp.symbols("t", real=True)
C = sp.Rational(1, 22000)
v = 4 * t**2 + 2 * t
i = sp.simplify(C * sp.diff(v, t))
i_mA = float(i.subs(t, 3)) * 2200
print("Persamaan arus:", i, "A")
print(f"Arus pada 3 s: {i_mA:.2f} mA")
```

```python
import sympy as sp

# 1. Deklarasi simbol dan persamaan tegangan
t = sp.symbols("t", real=True)
v = 5 * t**2
v_2s = v.subs(t, 2)

# 2. Hitung turunan pertama (dv/dt)
dv_dt = sp.diff(v, t)
laju_2s = dv_dt.subs(t, 2)

# 3. Hitung arus kapasitor i_C(t) = C * (dv/dt)
C = sp.Rational(1, 10000)  # C = 100 uF = 1/10000 F (pecahan eksak)
i_c = sp.simplify(C * dv_dt)
i_c_2s_mA = float(i_c.subs(t, 2)) * 1000

# --- PENAMPILAN HASIL UTAMA (PRINT) ---
print("=" * 40)
print("   HASIL KOMPUTASI SIMBOLIK SYMPY")
print("=" * 40)
print("Persamaan tegangan v(t)      :", v, "V")
print("Tegangan pada t = 2 s        :", v_2s, "V")
print("Persamaan dv/dt              :", dv_dt, "V/s")
print("Laju perubahan pada t = 2 s  :", laju_2s, "V/s")
print("Persamaan arus i_C(t)        :", i_c, "A")
print(f"Arus kapasitor pada t = 2 s  : {i_c_2s_mA:.2f} mA")
print("=" * 40)
```

```python
import sympy as sp

# ===================================================
# LATIHAN MAHASISWA - PERTEMUAN 2 (TURUNAN PYTHON)
# Pemrograman Komputer Lanjut - Teknik Elektro UWG
# ===================================================

# ---------------------------------------------------
# SOAL A: Turunan Polinomial
# ---------------------------------------------------
t = sp.symbols("t", real=True)

# 1. Definisi fungsi f(t)
f = 2 * t**3 - 4 * t**2 + 6 * t - 1

# 2. Hitung turunan pertama, kedua, dan evaluasi t = 2
df = sp.diff(f, t)  # f'(t)
d2f = sp.diff(f, t, 2)  # f''(t)
df_2 = df.subs(t, 2)  # f'(2)

# ---------------------------------------------------
# SOAL B: Arus Kapasitor
# ---------------------------------------------------
# C = 220 uF = 220/1000000 F (menggunakan pecahan eksak)
C = sp.Rational(220, 1000000)
v = 4 * t**2 + 2 * t

# 1. Hitung dv/dt dan arus i(t)
dv_dt = sp.diff(v, t)
i_t = C * dv_dt

# 2. Evaluasi arus pada t = 3 detik (dalam mA)
i_3_A = i_t.subs(t, 3)
i_3_mA = float(i_3_A) * 1000

# ===================================================
# MENAMPILKAN HASIL (PRINT FORMATTED)
# ===================================================
print("=" * 50)
print("     HASIL KOMPUTASI SIMBOLIK LATIHAN MAHASISWA")
print("=" * 50)

print("\n--- [ SOAL A ] ---")
print("Fungsi f(t)        :", f)
print("Turunan f'(t)      :", df)
print("Turunan f''(t)     :", d2f)
print("Nilai f'(2)        :", df_2)

print("\n--- [ SOAL B ] ---")
print("Kapasitansi C      :", C, "F (220 uF)")
print("Tegangan v(t)      :", v, "V")
print("Turunan dv/dt      :", dv_dt, "V/s")
print("Persamaan i(t)     :", i_t, "A")
print(f"Arus i(3)          : {i_3_mA:.2f} mA")

print("\n" + "=" * 50)
```
