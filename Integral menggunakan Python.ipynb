```python
import sympy as sp
t = sp.symbols("t", real=True)
f = 6 * t + 4
F_dasar = sp.integrate(f, t)
K = 5 - F_dasar.subs(t, 0)
F = sp.expand(F_dasar + K)
print("K =", K)
print(F)
print("F(0) =", F.subs(t, 0))
```

```python
import sympy as sp
t = sp.symbols("t", real=True)
f = 6 * t + 4
hasil = sp.integrate(f, (t, 0, 2))
F = sp.integrate(f, t)
cek = F.subs(t, 2) - F.subs(t, 0)
print("Integral:", hasil)
print("Pemeriksaan:", cek)
```

```python
import sympy as sp
t = sp.symbols("t", real=True)
P = 10 + 2 * t
E = sp.integrate(P, (t, 0, 5))
P_rata = E / 5
E_Wh = E / 3600
print(f"Energi = {float(E):.2f} J")
print(f"Energi = {float(E_Wh):.6f} Wh")
print(f"Rata-rata = {float(P_rata):.2f} W")
```

```python
import sympy as sp
t = sp.symbols("t", nonnegative=True)
tau = sp.symbols("tau", real=True)
C = sp.Rational(1, 10000)
i = sp.Rational(1, 1000)
dq = sp.integrate(i, (tau, 0, t))
v = sp.simplify(2 + dq / C)
print(v)
print("Tegangan:",v.subs(t, sp.Rational(1, 2)),"V")
```

```python
import sympy as sp
t = sp.symbols("t", real=True)
f = 8 * t + 3

# Menghitung antiturunan
F = sp.integrate(f, t)
print("Antiturunan:", F) 

# Menghitung integral tentu batas 0 sampai 2
hasil = sp.integrate(f, (t, 0, 2))
print("Integral 0-2:", hasil)
```

```python
import sympy as sp
t = sp.symbols("t", real=True)
P = 15 + 3 * t

# Integral batas 0 sampai 4 sekon
E = sp.integrate(P, (t, 0, 4))
P_rata = E / 4
E_Wh = E / 3600

print(f"Energi = {float(E):.2f} J")
print(f"Energi = {float(E_Wh):.6f} Wh")
print(f"Rata-rata = {float(P_rata):.2f} W")
```

```python
import sympy as sp
t = sp.symbols("t", nonnegative=True)
tau = sp.symbols("tau", real=True)

# Konversi: C = 200 uF = 2/10000 F, i = 2 mA = 2/1000 A
C = sp.Rational(2, 10000)
i = sp.Rational(2, 1000)

dq = sp.integrate(i, (tau, 0, t))
v = sp.simplify(1 + dq / C)

print("Persamaan Tegangan:", v)
print("vC(0.3) =", v.subs(t, sp.Rational(3, 10)), "V")
```
